```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
student_data = pd.read_csv("datasets/dp_dataset.csv")
student_data
```

```python
student_data["EnrollmentDate"] = pd.to_datetime(student_data["EnrollmentDate"])

student_data["Month"] = student_data["EnrollmentDate"].dt.strftime("%B")
```

```python
monthly_enrollment = student_data.groupby("Month")["StudentID"].count()
plt.figure(figsize=(10,5))
plt.plot(monthly_enrollment.index,monthly_enrollment.values,color='orange')
plt.title("Monthly Student Enrollments")
plt.xlabel("Month")
plt.ylabel("Number of Students")
plt.grid(True)
plt.show()
```

```python
monthly_revenue = student_data.groupby("Month")["Revenue"].sum()
plt.figure(figsize=(10,5))
plt.plot(monthly_revenue.index,monthly_revenue.values,)
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid(True)
plt.show()
```

```python
monthly_progress = student_data.groupby("Month")["CourseProgress"].mean()
plt.figure(figsize=(10,5))
plt.plot(monthly_progress.index, monthly_progress.values,color='gold')
plt.title("Average Course Progress by Month")
plt.xlabel("Month")
plt.ylabel("Course Progress")
plt.grid(True)
plt.show()
```

```python
monthly_attendance = student_data.groupby("Month")["AttendancePercentage"].mean()
plt.figure(figsize=(10,5))
plt.plot(monthly_attendance.index, monthly_attendance.values)
plt.title("Average Attendance Percentage by Month")
plt.xlabel("Month")
plt.ylabel("Attendance %")
plt.grid(True)
plt.show()
```

```python
course_count = student_data["Course"].value_counts()
plt.figure(figsize=(10,5))
plt.bar(course_count.index, course_count.values)
plt.title("Number of Students in Each Course")
plt.xlabel("Course")
plt.ylabel("Students")
plt.show()
```

```python
course_revenue = student_data.groupby("Course")["Revenue"].sum()
plt.figure(figsize=(10,5))
plt.bar(course_revenue.index, course_revenue.values,color='red')
plt.title("Revenue Generated by Each Course")
plt.xlabel("Course")
plt.ylabel("Revenue")
plt.show()
```

```python
print(course_revenue.idxmax())   # Course name
print(course_revenue.max())      # Revenue amount
```
