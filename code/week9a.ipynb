```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

```python
df = pd.read_csv("datasets/student_dropout_dataset.csv")
display(df)
```

```python
print("Shape:", df.shape)
print("Columns:")
print(df.columns.tolist())

print("Missing values:")
print(df.isnull().sum())

print("Target distribution:")
print(df["DropoutRisk"].value_counts())

print("Basic statistics:")
display(df.describe())
```

```python
X = df[
    [
        "Attendance",
        "AssignmentCompletion",
        "PreviousMarks",
        "MissedClasses",
        "LateSubmissions"
    ]
]

y = df["DropoutRisk"]

print("X (features):")
display(X.head())

print("y (target):")
display(y.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
print("Logistic Regression model trained successfully!")
```

```python
y_pred = model.predict(X_test)

results = X_test.copy()
results["Actual"] = y_test.values
results["Predicted"] = y_pred

display(results)
```

```python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print(f"Accuracy : {accuracy:.2f}")
print(f"Precision: {precision:.2f}")
print(f"Recall   : {recall:.2f}")
print(f"F1-Score : {f1:.2f}")

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("Classification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["Not at risk", "At risk"]
))
```

## 9. Why is this a Binary Classification Problem?

This is a **binary classification** problem because the target variable has only two possible classes:

- `0` → Not at risk
- `1` → At risk

The model must choose one of these two categories for each student.

## 10. Why Use Logistic Regression?

Logistic Regression is suitable because:

1. The target is binary.
2. It predicts the probability of a student being at risk.
3. It is simple and interpretable.
4. It works well as a baseline classification model.
5. The model can show how each feature contributes to the prediction.

## 11. Which Factors are Related to Higher Dropout Risk?

From this dataset, students with:

- lower attendance,
- lower assignment completion,
- lower previous marks,
- more missed classes, and
- more late submissions

generally appear more likely to be at risk.


## 12. Test the Model with a New Student

Example new student:

- Attendance = 60
- Assignment Completion = 50
- Previous Marks = 48
- Missed Classes = 7
- Late Submissions = 6

```python
new_student = pd.DataFrame({
    "Attendance": [60],
    "AssignmentCompletion": [50],
    "PreviousMarks": [48],
    "MissedClasses": [7],
    "LateSubmissions": [6]
})

prediction = model.predict(new_student)[0]
probability = model.predict_proba(new_student)[0][1]

if prediction == 1:
    print("Prediction: AT RISK of dropping out")
else:
    print("Prediction: NOT AT RISK of dropping out")

print(f"Probability of dropout risk: {probability:.2%}")
```

A Logistic Regression model was trained to predict student dropout risk using attendance,
assignment completion, previous marks, missed classes, and late submissions.

The model classifies students into:
