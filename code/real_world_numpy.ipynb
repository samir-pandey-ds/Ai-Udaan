```python
import pandas as pd
import numpy as np

data = pd.read_csv("datasets/daraz_dataset.csv")
data
```

```python
data.head()
data.shape
data.columns
data.dtypes
data["Age"]
data["Age"] = pd.to_numeric(data["Age"],errors = "coerce")
```

```python
#shortcutt for chnage dtypes all at a time

numeric_columns = [
    "Age",
    "Quantity",
    "UnitPrice",
    "DiscountPercent",
    "TotalAmount",
    "Rating"
]
for  column in numeric_columns:
    data[column] = pd.to_numeric(data[column],errors="coerce")
```

```python
data.isnull().sum()
data["Quantity"] = data["Quantity"].fillna(data["Quantity"].mean())
```

```python
features  = ["Age","Quantity","UnitPrice" ,"DiscountPercent"]
target = "TotalAmount"

#super features store garne X , target store garne y bananunu

x = data[features]
y = data[target]

type(x)
type(y)

x = x.to_numpy()
y = y.to_numpy()

type(x)
type(y)

x.shape
x.ndim
x.dtype
x.nbytes
```
