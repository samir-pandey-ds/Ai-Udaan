```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
index = [1,2,3,4] #x(horizontal)
price =[1000,2000,3000,4300] #y(vertical)
plt.plot(index,price)
plt.show()
```

```python
dd = pd.read_csv("datasets/daraz_dataset.csv")
dd
```

```python
#bargraph
category = dd.groupby("Category")["Product"].count()
plt.figure(figsize=(18,6))
plt.bar(category.index,category.values,color="green",edgecolor="blue")
plt.xlabel("categories",fontsize=20)
plt.ylabel("Products",fontsize=20)
plt.title("Products and Category bargraph")
plt.grid(axis="y",linestyle="--",alpha=0.3)
```

```python
category = dd.groupby("Category")["Product"].count()
category
```

```python
#piechart
color  = [
    "green",
    "brown",
    "blue",
    "red",
    "orange",
    "purple",
    "pink"
]
plt.figure(figsize=(12,5))    #size change garxa
plt.pie(category.values,labels = category.index,colors=color,shadow=True,startangle=90,explode = (0.1,0,0,0,0,0,0.1)) #category.values le pie chart matra ani labels le name pani dekhauxa
plt.title("Product distribution by category")
plt.show()
```

```python
gender = dd.groupby("Gender")["Product"].count()
gender
```

```python
plt.pie(gender.values,labels=gender.index,colors=["pink","green"])
plt.title("Products by Gender")
```

```python
#histogram (data ko distribution garxa)
bins = [0,5000,10000,20000,50000,80000]
plt.hist(dd["UnitPrice"],bins=bins,color="green",edgecolor="blue",alpha=0.5)
plt.title("Distribution of UnitPrice")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")
```

```python
#Scatter
data = {
    "Age" : [20,22,24,26,28,30],
    "TotalAmount" : [1000,1200,1500,2000,3000,32000]
    }
dfd =pd.DataFrame(data)
dfd
plt.scatter(dfd["Age"],dfd["TotalAmount"],color = "green")
plt.xlabel("Age")
plt.ylabel("TotalAmount")
plt.grid(True)
plt.title("Age Vs TotalAmount")
```
